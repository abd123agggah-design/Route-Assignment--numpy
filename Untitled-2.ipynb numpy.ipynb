```python
import numpy as np

arr = np.random.randint(1, 101, (3, 3))
normalized = (arr - arr.min()) / (arr.max() - arr.min())

print(arr)
print(normalized)
```

```python
import numpy as np

arr = np.random.randint(1, 100, 10)
x = float(input("X: "))
closest = arr[np.abs(arr - x).argmin()]

print("Array:", arr)
print("Closest to", x, "is", closest)
```

```python
import numpy as np

arr = np.random.randint(1, 100, (5, 5))
second_largest = np.sort(arr, axis=0)[-2]

print(arr)
print("Second largest in each column:", second_largest)
```

```python
import numpy as np

arr = np.random.randint(1, 100, (6, 4))
even_rows = np.where(arr.sum(axis=1) % 2 == 0)[0]

print(arr)
print("Rows with even sum:", even_rows)
```

```python
import numpy as np

arr = np.random.randint(1, 51, 30)
arr[(arr % 3 == 0) | (arr % 5 == 0)] = -1

print(arr)
```

```python
import numpy as np

arr = np.random.randint(1, 100, (3, 3))
result = arr - arr.mean(axis=1, keepdims=True)

print(arr)
print(result)
```
